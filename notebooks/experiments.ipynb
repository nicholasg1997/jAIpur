```python
model_env = DummyVecEnv([lambda: JaipurEnv(seed=42)])
model = MultiOutputPPO.load("/Users/nickgault/PycharmProjects/jAIpur/jaipur_rl/training/models/Jaipur_25.0M_scheduler_2025-06-15_14-41-20.zip", env=model_env)
env = model.get_env()
trained_agent = model
done = False
truncated = False
p1_scores = []
p2_scores = []

for i in range (50):
    obs = env.reset()
    print(f"Game {i+1} Start")
    for i in range(200):
        #print(f"Step {i+1}")
        action, _ = trained_agent.predict(obs, deterministic=True)
        #print(f"Action: {action}")
        obs, reward, done, truncated, info = env.step(action)
        #env.render()
        if done or truncated:
            print("Game Over")
            #print(info)
            p1_score = info['P1_Final_Score']
            p1_scores.append(p1_score)
            p2_score = info['P2_Final_Score']
            p2_scores.append(p2_score)
            break

print(f"Player 1 Score: {sum(p1_scores)/50}, Player 2 Score: {sum(p2_scores)/50}")
```
